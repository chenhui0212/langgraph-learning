## 子图

#### 检查点

```python
from typing import TypedDict

from langgraph.checkpoint.memory import InMemorySaver
from langgraph.graph import StateGraph, START, END


# ===== 子图：文本清洗 + 润色 =====
class SubgraphState(TypedDict):
    raw_text: str
    clean_text: str
    final_text: str

def clean(state: SubgraphState):
    return {"clean_text": state["raw_text"].strip()}

def polish(state: SubgraphState):
    return {"final_text": f"{state['clean_text']}（已润色）"}

sub_builder = StateGraph(SubgraphState)
sub_builder.add_node("clean", clean)
sub_builder.add_node("polish", polish)
sub_builder.add_edge(START, "clean")
sub_builder.add_edge("clean", "polish")
sub_builder.add_edge("polish", END)
subgraph = sub_builder.compile()


# ===== 父图 =====
class ParentState(TypedDict):
    raw_text: str
    final_text: str

builder = StateGraph(ParentState)
builder.add_node("subgraph_node", subgraph)
builder.add_edge(START, "subgraph_node")
builder.add_edge("subgraph_node", END)
graph = builder.compile(checkpointer=InMemorySaver())

config = {"configurable": {"thread_id": "8-2-1"}}
result = graph.invoke({"raw_text": "  LangGraph 子图示例  "}, config=config)
print(result)
```

```python
from rich.pretty import pprint

# 查看状态历史（不包含完整子图状态历史）
histories = list(graph.get_state_history(config=config))
pprint(histories)


# 获取调用子图的状态，再获取对应子图的状态历史
parent_config = histories[-2].config
pprint(graph.get_state(config=parent_config, subgraphs=True))
```
