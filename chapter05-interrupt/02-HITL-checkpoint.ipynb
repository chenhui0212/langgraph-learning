## 中断

#### 串行中断检查点

```python
from typing import TypedDict

from langgraph.checkpoint.memory import InMemorySaver
from langgraph.graph import StateGraph, START, END
from langgraph.types import Command
from rich.pretty import pprint


class State(TypedDict):
    username: str
    age: int

def a(state: State):
    name = interrupt("请输入您的姓名")
    age = interrupt("请输入您的年龄")
    return {
        "username": name,
        "age": age
    }


builder = StateGraph(State)
builder.add_node(a)
builder.add_edge(START, "a")
builder.add_edge("a", END)
graph = builder.compile(checkpointer=InMemorySaver())

config = {"configurable": {"thread_id": "5-2-1"}}
result = graph.invoke({}, config=config)
print(result)

pprint(list(graph.get_state_history(config=config)))
```

```python
result = graph.invoke(Command(resume="张三"), config=config)
print(result)
pprint(list(graph.get_state_history(config=config)))
```

```python
result = graph.invoke(Command(resume=18), config=config)
print(result)
pprint(list(graph.get_state_history(config=config)))
```

#### 并行中断检查点

```python
from langgraph.graph import StateGraph, START, END
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.types import Command, interrupt
from typing import TypedDict
from rich.pretty import pprint


class State(TypedDict):
    username: str
    age: int

def a(state: State):
    name = interrupt("请输入您的姓名")
    return {
        "username": name
    }

def b(state: State):
    age = interrupt("请输入您的年龄")
    return {
        "age": int(age)
    }


builder = StateGraph(State)
builder.add_node(a)
builder.add_node(b)
builder.add_edge(START, "a")
builder.add_edge(START, "b")
builder.add_edge("a", END)
builder.add_edge("b", END)
graph = builder.compile(checkpointer=InMemorySaver())

config = {"configurable": {"thread_id": "5-2-2"}}
result = graph.invoke({}, config=config)
print(result)
pprint(list(graph.get_state_history(config=config)))
```

```python
resume_map = {}
for prompt in result['__interrupt__']:
    resume_map[prompt.id] = input(prompt.value)
result = graph.invoke(Command(resume=resume_map), config=config)
print(result)
pprint(list(graph.get_state_history(config=config)))
```
