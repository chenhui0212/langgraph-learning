#### [State Reducer](https://docs.langchain.com/oss/python/langgraph/graph-api#reducers)

##### 自定义 Reducer 函数

```python
from typing import TypedDict, Annotated

# left:  当前已累计的状态值
# right: 本次待合并的新值
def my_reducer(left: list[str], right: list[str]) -> list[str]:
    return left + right


class State(TypedDict):
    logs: Annotated[list[str], my_reducer]
    cur_id: str
```

#### 常用内置 Reducer 函数

##### operator.add

```python
from operator import add

print(f"{add(1, 2) = }")
print(f"{add([1, 2], [3, 4]) = }")
print(f"{add(['a', 'b'], ['c']) = }")
```

##### langgraph.graph.message.add_messages

```python
from langgraph.graph.message import add_messages
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

left = [
    SystemMessage("你是一个天气助手", id="1"),
    HumanMessage("你好", id="2"),
    AIMessage("你好，我是AI", id="3"),
]

right = [
    HumanMessage("你好, 我叫张三", id="2"),
    AIMessage("你好, 张三", id="3"),
    HumanMessage("请问今天天气如何", id="4"),
]

merged = add_messages(left, right)
for msg in merged:
    print(msg)
```
